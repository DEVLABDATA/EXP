```python
# Experiment 10: Text Classification using Naive Bayes and SVM
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
docs = ['The team won the championship after a thrilling final match',
'The striker scored a hat-trick in yesterday football game',
'The new smartphone features a faster processor and better camera',
'Artificial intelligence is transforming the technology industry',
'The election results were announced amid tight security',
'The senator proposed a new policy on healthcare reform',
'The cricket team practiced hard for the upcoming tournament',
'The latest laptop comes with improved battery life and AI chip']
labels = ['Sports', 'Sports', 'Technology', 'Technology', 'Politics', 'Politics', 'Sports', 'Technology']
x, xt, y, yt = train_test_split(docs, labels, test_size=0.25, random_state=1)
for name, clf in [('Naive Bayes', MultinomialNB()), ('SVM', SVC(kernel='linear'))]:
    model = make_pipeline(TfidfVectorizer(), clf).fit(x, y)
    pred = model.predict(xt)
    print(name, 'Predictions:', list(pred))
    print(name, 'Accuracy:', accuracy_score(yt, pred))
```
