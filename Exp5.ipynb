```python
# Experiment 5: CFG Parsing
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N | Det N PP
VP -> V NP
PP -> P NP
Det -> 'the' | 'a'
N -> 'man' | 'dog' | 'telescope'
V -> 'saw'
P -> 'with'
""")

sentence = 'the man saw a dog with the telescope'.split()
print('Parse Trees for:', ' '.join(sentence))

for tree in nltk.ChartParser(grammar).parse(sentence):
    print(tree)
    tree.pretty_print()
```
