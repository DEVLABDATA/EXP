```python
# Experiment 7: Word Sense Disambiguation
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize
for sentence in ['He deposited money in the bank', 'They sat on the river bank']:
    sense = lesk(word_tokenize(sentence), 'bank')
    print(sentence, '->', sense.definition() if sense else 'No sense found')
```
