```python
# Experiment 2: Regular Expressions
import re
text = 'Contact us at student@example.com or call 9876543210.'
print('Emails:', re.findall(r'[\w.-]+@[\w.-]+\.\w+', text))
print('Phone numbers:', re.findall(r'\b\d{10}\b', text))
print('Words starting with N:', re.findall(r'\bN\w*', 'NLP needs natural language tools'))
```
