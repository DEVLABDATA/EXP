```python
# Experiment 11: Information Retrieval using TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import precision_score, recall_score, f1_score
docs = ['Natural language processing enables computers to understand human language.',
'Information retrieval systems retrieve relevant documents for a user query.',
'An inverted index maps each term to documents containing it.',
'TF-IDF measures the importance of a term in a document collection.',
'Search engines use crawling indexing and ranking to answer user queries.',
'Machine learning models classify text documents into categories.',
'Cosine similarity ranks documents by comparing document and query vectors.',
'Precision and recall evaluate information retrieval systems.']
query = 'evaluate information retrieval systems documents'
v = TfidfVectorizer()
X = v.fit_transform(docs + [query])
scores = cosine_similarity(X[-1], X[:-1]).flatten()
ranked = scores.argsort()[::-1][:3]
print('Query:', query)
print('\nRanked Results:')
for rank, idx in enumerate(ranked, 1):
    print(f'Rank {rank}: Doc {idx+1} Score = {scores[idx]:.4f}')
actual = [int(i in (1, 7)) for i in range(len(docs))]
retrieved = [int(i in ranked[:2]) for i in range(len(docs))]
print('\nPrecision =', f'{precision_score(actual, retrieved):.2f}',
      'Recall =', f'{recall_score(actual, retrieved):.2f}',
      'F1-score =', f'{f1_score(actual, retrieved):.2f}')
```
