```python
# Experiment 8: Named Entity Recognition
import spacy
# Install model once if needed: python -m spacy download en_core_web_sm
nlp = spacy.load('en_core_web_sm')
doc = nlp('Sundar Pichai, the CEO of Google, visited Bangalore.')
for ent in doc.ents:
    print(ent.text, '->', ent.label_)
```
