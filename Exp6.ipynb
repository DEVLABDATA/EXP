```python
# Experiment 6: PCFG Parsing
import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> 'John' [0.5] | Det N [0.5]
VP -> V NP [1.0]
Det -> 'a' [1.0]
N -> 'dog' [1.0]
V -> 'saw' [1.0]
""")

sentence = 'John saw a dog'.split()
print('Most probable parse for:', ' '.join(sentence))

for tree in nltk.ViterbiParser(grammar).parse(sentence):
    print(tree)
    print('Probability:', tree.prob())
```
