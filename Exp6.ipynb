```python
# Experiment 6: PCFG Parsing
import nltk
grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [1.0]
VP -> V NP [1.0]
Det -> 'the' [0.5] | 'a' [0.5]
N -> 'student' [0.5] | 'book' [0.5]
V -> 'reads' [1.0]
""")
for tree in nltk.ViterbiParser(grammar).parse('the student reads a book'.split()):
    print(tree)
    print('Probability:', tree.prob())
```
