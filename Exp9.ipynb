```python
# Experiment 9: Sentiment Analysis
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
reviews = ['Movie was fantastic and thrilling', 'I loved the acting and story',
'Wonderful and heartwarming film', 'Best movie, brilliant',
'Plot was boring and acting terrible', 'I hated this movie',
'Poor direction and dull story', 'Worst film, disappointing']
labels = ['positive'] * 4 + ['negative'] * 4
a, b, y1, y2 = train_test_split(reviews, labels, test_size=0.25, random_state=42)
v = TfidfVectorizer()
a, b = v.fit_transform(a), v.transform(b)
model = MultinomialNB().fit(a, y1)
pred = model.predict(b)
print('Accuracy:', accuracy_score(y2, pred))
print('Classification Report:\n', classification_report(y2, pred))
new = ['The film was amazing and beautifully shot', 'This was a terrible and boring movie']
for review, sentiment in zip(new, model.predict(v.transform(new))):
    print('\nReview:', review)
    print('Predicted Sentiment:', sentiment)
```
