```python
# Experiment 4: Morphological Analysis and FSA
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ['playing', 'played', 'plays', 'happily']
for word in words:
    print(word, '-> stem:', stemmer.stem(word))
for word in words:
    print(word, 'accepted:', word.endswith('ing'))
```
