```python
# Experiment 4: Morphological Analysis and FSA
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ['playing', 'played', 'plays', 'player', 'fastest', 'quickly', 'cat']

print('Morphological Analysis:\n')
for word in words:
    print(word, '-> stem:', stemmer.stem(word))

print('\nFSA Word Recognition:\n')
test_words = ['play', 'playing', 'played', 'plays', 'playist', 'run']
for word in test_words:
    if word == 'play' or word.endswith(('ing', 'ed', 's')):
        print(word, '-> ACCEPTED')
    else:
        print(word, '-> REJECTED')
```
