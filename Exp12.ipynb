```python
# Experiment 12: Extractive Text Summarizer
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from heapq import nlargest

def summarize(text, n=2):
    sentences = sent_tokenize(text)
    stop = set(stopwords.words('english'))
    freq = {}
    for word in word_tokenize(text.lower()):
        if word.isalpha() and word not in stop:
            freq[word] = freq.get(word, 0) + 1
    if not freq:
        return ''
    maximum = max(freq.values())
    freq = {w: f / maximum for w, f in freq.items()}
    scores = {}
    for sentence in sentences:
        for word in word_tokenize(sentence.lower()):
            if word in freq:
                scores[sentence] = scores.get(sentence, 0) + freq[word]
    best = nlargest(n, scores, key=scores.get)
    return ' '.join(s for s in sentences if s in best)

paragraph = "Natural language processing (NLP) is a branch of artificial intelligence that helps computers understand, interpret and manipulate human language. NLP combines computational linguistics with statistical, machine learning and deep learning models. These technologies enable computers to process human language in the form of text or voice data and to understand its full meaning. NLP is used in applications such as chatbots, machine translation, sentiment analysis and information retrieval systems. The growth of digital text data has made NLP increasingly important for businesses and researchers alike."
summary = summarize(paragraph)
print('Original Paragraph:\n', paragraph)
print('\nOriginal Length:', len(paragraph.split()), 'words')
print('\nGenerated Summary:\n', summary)
print('\nSummary Length:', len(summary.split()), 'words')
```
